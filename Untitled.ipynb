```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
from PIL import Image

# Scikit-learn utilities
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# TensorFlow / Keras for image processing and augmentation
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print("TensorFlow Version:", tf.__version__)
print("All base packages imported successfully.")
```

### LOAD DATASET

```python
df_tabular = pd.read_csv('crop_yield_dataset.csv')
```

```python
df_tabular
```

```python
#Define Target Variable: Crop_Stress (1 if yield is in lower 25th percentile, else 0)
yield_threshold = df_tabular['Crop_Yield'].quantile(0.25)
df_tabular['Crop_Stress'] = np.where(df_tabular['Crop_Yield'] <= yield_threshold, 1, 0)
```

```python
df_tabular
```

### Soil Dataset Preprocession

```python
#One-Hot Encoding for Categorical Columns
df_encoded = pd.get_dummies(df_tabular, columns=['Soil_Type', 'Crop_Type'], drop_first=True)
```

```python
#Feature and Target Separation
feature_cols = [col for col in df_encoded.columns if col not in ['Date', 'Crop_Yield', 'Crop_Stress']]
X_tabular = df_encoded[feature_cols]
y_tabular = df_encoded['Crop_Stress']
```

```python
# Partition Data: 70% Train, 15% Validation, 15% Test
X_train_tab, X_temp_tab, y_train_tab, y_temp_tab = train_test_split(
    X_tabular, y_tabular, test_size=0.30, random_state=42, stratify=y_tabular
)
X_val_tab, X_test_tab, y_val_tab, y_test_tab = train_test_split(
    X_temp_tab, y_temp_tab, test_size=0.50, random_state=42, stratify=y_temp_tab
)
```

```python
# Feature Normalization
scaler = StandardScaler()
X_train_tab_scaled = scaler.fit_transform(X_train_tab)
X_val_tab_scaled = scaler.transform(X_val_tab)
X_test_tab_scaled = scaler.transform(X_test_tab)

print(f"Tabular Train Shape: {X_train_tab_scaled.shape}")
print(f"Tabular Validation Shape: {X_val_tab_scaled.shape}")
print(f"Tabular Test Shape: {X_test_tab_scaled.shape}")
```

### PlantVillage Image Pipeline & Data Augmentation

```python
# Configuration Parameters
IMG_HEIGHT, IMG_WIDTH = 224, 224
BATCH_SIZE = 32
PLANTVILLAGE_DIR = 'plantvillage dataset\color'

# Image Data Generator with Augmentation for Training
train_datagen = ImageDataGenerator(
    rescale=1.0/255.0,
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,
    validation_split=0.20  # 80% Train, 20% Validation split
)

# Data Generator for Test/Validation (Rescaling only)
test_datagen = ImageDataGenerator(rescale=1.0/255.0)

# Load Training Images
train_generator = train_datagen.flow_from_directory(
    PLANTVILLAGE_DIR,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    shuffle=True
)

# Load Validation Images
val_generator = train_datagen.flow_from_directory(
    PLANTVILLAGE_DIR,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    shuffle=False
)

print(f"Total Disease/Health Classes Found: {len(train_generator.class_indices)}")
```

### Saving Processed Dataset

```python
# Create output directory for processed files
os.makedirs('data/processed', exist_ok=True)

# Save processed tabular arrays
np.save('data/processed/X_train_tab.npy', X_train_tab_scaled)
np.save('data/processed/X_val_tab.npy', X_val_tab_scaled)
np.save('data/processed/X_test_tab.npy', X_test_tab_scaled)

np.save('data/processed/y_train_tab.npy', y_train_tab.to_numpy())
np.save('data/processed/y_val_tab.npy', y_val_tab.to_numpy())
np.save('data/processed/y_test_tab.npy', y_test_tab.to_numpy())

# Save feature column names for SHAP interpretability later
pd.Series(feature_cols).to_csv('data/processed/feature_names.csv', index=False)

print("Processed tabular dataset successfully saved to 'data/processed/'.")
```
